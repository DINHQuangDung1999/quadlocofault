# Evaluation Grid Table

Load the evaluation grid CSV and display it as a formatted table with a few pivoted comparison views.

```python
from pathlib import Path

import pandas as pd

pd.set_option("display.max_rows", 100)
pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 160)

CSV_PATH = Path("/home/dung-admin/quadloco_ws/quadlocofault/logs/rsl_rl/unitree_go2_rough_gcn/2026-06-11_03-18-37/eval_grid_results.csv")
CSV_PATH
```

```python
df = pd.read_csv(CSV_PATH)
df = df.sort_values(["terrain_family", "fault_coef"]).reset_index(drop=True)
df
```

```python
numeric_cols = [
    "mean_reward",
    "abs_lin_x_err",
    "abs_lin_y_err",
    "abs_lin_xy_err",
    "abs_yaw_err",
]

display_df = df.copy()
display_df[numeric_cols] = display_df[numeric_cols].round(4)
display_df
```

```python
styled = (
    display_df.style
    .format({col: "{:.4f}" for col in numeric_cols})
    .background_gradient(subset=["mean_reward"], cmap="Greens")
    .background_gradient(subset=["abs_lin_x_err", "abs_lin_y_err", "abs_lin_xy_err", "abs_yaw_err"], cmap="Reds_r")
)
styled
```

## Pivot: Mean Reward

```python
reward_pivot = df.pivot(index="terrain_family", columns="fault_coef", values="mean_reward").round(4)
reward_pivot
```

## Pivot: Absolute XY Tracking Error

```python
xy_err_pivot = df.pivot(index="terrain_family", columns="fault_coef", values="abs_lin_xy_err").round(4)
xy_err_pivot
```

## Pivot: Absolute Yaw Tracking Error

```python
yaw_err_pivot = df.pivot(index="terrain_family", columns="fault_coef", values="abs_yaw_err").round(4)
yaw_err_pivot
```
